### 5: Leyendo banco de datos

```python
from sqlalchemy import create_engine, MetaData, Table, inspect, text
```

```python
engine = create_engine('sqlite:///:memory:') #crear una base de datos sqlite en memoria local
```

#### escribiendo en un banco de datos

```python
import pandas as pd
```

```python
archivo = 'clientes_banco.csv'
```

```python
datos = pd.read_csv(archivo)
```

```python
datos.head()
```

```python
datos.to_sql('clientes', engine, index=False)
```

```python
inspector = inspect(engine)#revisa todo lo que hay en engine y lo almaceno
```

```python
print(inspector.get_table_names())
```

#### Leyendo una consulta SQL

```python
query = 'SELECT * FROM clientes WHERE Categoria_de_renta = "Empleado";'
```

```python
empleados = pd.read_sql(sql=text(query), con = engine.connect())
```

```python
type(empleados)
```

```python
empleados
```

```python
#guardar el dataframe empleados en un archivo sql
empleados.to_sql('empleados', con = engine.connect(), index=False)
```

```python
#leer la tabla completa empleados desde la base de datos sql
pd.read_sql_table('empleados', con = engine.connect())
```

```python
#realizo la lectura de la tabla empleados con algunas columnas
pd.read_sql_table('empleados', con = engine.connect(), columns=['ID_Cliente', 'Grado_estudio', 'Rendimiento_anual'])
```

#### Actualizando un banco de datos

```python
query = 'SELECT * FROM clientes'
```

```python
pd.read_sql(query, con = engine.connect())
```

```python
#metodo que brinda informacion sobre errores en SQLAlchemy
from sqlalchemy.exc import SQLAlchemyError
```

```python
#eliminar un cliente de la base de datos clientes
query = "DELETE FROM clientes WHERE ID_Cliente=5008804;"

try:
    r_set = engine.connect().execute(text(query))
except SQLAlchemyError as e:
    print(e)
else:
    print("#Registros borrados: ", r_set.rowcount)
```

```python
pd.read_sql('clientes', con = engine.connect())
```

```python
#actualizar un cliente de la base de datos clientes
query = "UPDATE clientes SET Grado_estudio='Nivel superior' WHERE ID_Cliente=5008808;"

try:
    r_set = engine.connect().execute(text(query))
except SQLAlchemyError as e:
    print(e)
else:
    print("#Registros actualizados: ", r_set.rowcount)
```

```python
pd.read_sql('clientes', con = engine.connect())
```

```python
inspector = inspect(engine)
print(inspector.get_table_names())
```
