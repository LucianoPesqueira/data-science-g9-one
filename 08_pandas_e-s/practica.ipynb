```python
datos = pd.read_csv('datos_sus.csv', encoding='ISO-8859-1', sep=';', skiprows=3, skipfooter=9, engine='python')
datos.head()
```

### Desafio: manipulando una base de datos
archivo: clientes_banco_practica.csv

```python
from sqlalchemy import create_engine, text, inspect
from sqlalchemy.exc import SQLAlchemyError
import pandas as pd
```

```python
#1 --------- Crear la base de datos local con la biblioteca SQLAlchemy. -----------
engine = create_engine('sqlite:///:memory:')
```

```python
#2 --------- Escribir los datos del archivo CSV en esta base de datos local. -----------
archivo = 'clientes_banco_practica.csv' #almaceno en variable el nombre del archivo
datos = pd.read_csv(archivo) #leo el archivo csv y almaceno el dataframe en datos
datos.to_sql('clientes', engine, index=False) #escribo el df en la base de datos
```

```python
query = "SELECT * FROM clientes WHERE Categoria_de_renta='Empleado';"
clientes = pd.read_sql(sql=text(query), con=engine.connect())
clientes
```

```python
#3 --------- Realizar tres actualizaciones en la base de datos: -----------
#- Actualizar el registro del cliente ID 6840104 cuyo rendimiento anual cambió a 300000.
query = "UPDATE clientes SET Rendimiento_anual=300000 WHERE ID_Cliente=6840104"

try:
    r_set = engine.connect().execute(text(query))
except SQLAlchemyError as e:
    print(e)
else:
    print(f"Registros actualizados: {r_set.rowcount}")
```

```python
query = "SELECT * FROM clientes WHERE ID_Cliente=6840104;"
pd.read_sql(sql=text(query), con= engine.connect())
```

```python
#- Eliminar el registro de cliente ID 5008809, 
# ya que esta persona ya no tiene cuenta en la institución financiera.

query = "DELETE FROM clientes WHERE ID_Cliente=5008809"

try:
    r_set = engine.connect().execute(text(query))
except SQLAlchemyError as e:
    print(e)
else:
    print(f"Registros actualizados: {r_set.rowcount}")
```

```python
query = "SELECT * FROM clientes WHERE ID_Cliente=5008809;"
pd.read_sql(sql=text(query), con= engine.connect())
```
