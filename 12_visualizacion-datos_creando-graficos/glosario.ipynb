### 01- Conociendo la biblioteca Matplotlib

```python
#----------Conociendo los datos----------
import pandas as pd

df = pd.read_csv('./archivos/inmigrantes_canada.csv')
df
```

```python
df.info()
```

```python
#establezco la columna pais como indice
df.set_index('Pais', inplace=True)
df
```

```python
#lista de años
anos = list(map(str,range(1980, 2014)))
anos
```

```python
#cantidad de inmigrantes de colombia en cada año
colombia = df.loc['Colombia', anos]
colombia
```

```python
col_dict = {'Año': colombia.index.tolist(),
            'Inmigrantes':colombia.values.tolist()}
```

```python
#ya tengo el dataframe con los datos de colombianos que emigraron a canada por año
datos_col = pd.DataFrame(col_dict)
datos_col.tail()
```

```python
#----------Creando el primer grafico con matplotlib----------
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes']);
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2015']);
plt.show()
```

```python
#----------Añadiendo titulos y rotulos----------

plt.figure(figsize=(8,4))
plt.plot(datos_col['Año'], datos_col['Inmigrantes']);
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2015']);
plt.title('Inmigracion de Colombianos hacia canadá')
plt.xlabel('Año')
plt.ylabel('Nro de Inmigrantes')
plt.show() #siempre debe ser el ultimo comando
```

### Desafio: Comparando tendencias migratorias
* Crear un gráfico de líneas comparando los números de inmigrantes de Brasil y Argentina a Canadá

```python
#creo un dataframe con los datos de brasil y argentina
df_comparacion = df.loc[['Brasil', 'Argentina'], anos]
df_comparacion
```

```python
#tranformo el dataframe las filas por columnas y las columnas por filas

df_comparacion = df_comparacion.T
df_comparacion.head()
```

```python
#grafico la comparacion 

plt.figure(figsize=(8,4))
plt.plot(df_comparacion['Brasil'], label='Brasil');
plt.plot(df_comparacion['Argentina'], label='Argentina');
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010', '2015']);
plt.title('Comparacion de inmigracion de Brasil y Argentina a Canadá')
plt.xlabel('Año')
plt.ylabel('Nro de Inmigrantes')
plt.legend()
plt.show() #siempre debe ser el ultimo comando
```

### 02- Creando figuras con Matplotlib

```python
#----------Creando una figura----------

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'])
ax.xaxis.set_major_locator(plt.MultipleLocator(5))#remplaza el xticks de los años
plt.show()
```

```python
#----------Añadiendo titulos y rotulos en la figura----------

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'])
ax.xaxis.set_major_locator(plt.MultipleLocator(5))#remplaza el xticks de los años
ax.set_title('Inmigracion de Colombianos hacia canadá\n1980 - 2013')
ax.set_xlabel('Año')
ax.set_ylabel('Nro de Inmigrantes')
plt.show()
```

```python
#----------Creando subplots en una direccion----------

fig, axs = plt.subplots(1, 2, figsize =(10,3)) #una fila con dos graficas

axs[0].plot(datos_col['Año'], datos_col['Inmigrantes'])
axs[0].set_title('Inmigracion de Colombianos hacia canadá\n1980 - 2013')
axs[0].set_xlabel('Año')
axs[0].set_ylabel('Nro de Inmigrantes')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))#años en 5 en 5
axs[0].grid()

axs[1].boxplot(datos_col['Inmigrantes'])
axs[1].set_title('Boxplot de la inmigración de colombianos\nhacia Canadá 1980 - 2013')
axs[1].set_xlabel('Colombia')
axs[1].set_ylabel('Nro de Inmigrantes')
axs[1].grid()

plt.show()
```

```python
#----------Creando subplots en dos direcciones #1----------

fig, axs = plt.subplots(2, 2, figsize = (10, 6))
fig.subplots_adjust(hspace=0.5, wspace=0.5)#ajusto la dimension horizontal high y width

axs[0,0].plot(df.loc['Colombia', anos])
axs[0,0].set_title('Colombia')

axs[0,1].plot(df.loc['Brasil', anos])
axs[0,1].set_title('Brasil')

axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Perú', anos])
axs[1,1].set_title('Perú')

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))
    ax.set_xlabel('Año')
    ax.set_ylabel('Nro de Inmigrantes')

plt.show()
```

```python
#----------Creando subplots en dos direcciones #2----------

fig, axs = plt.subplots(2, 2, figsize = (10, 6))
fig.subplots_adjust(hspace=0.5, wspace=0.5)#ajusto la dimension horizontal high y width
fig.suptitle('Inmigracion de los 4 mayores paises sudamericanoshacia Canadá\n desde 1980 a 2013')

axs[0,0].plot(df.loc['Colombia', anos])
axs[0,0].set_title('Colombia')

axs[0,1].plot(df.loc['Brasil', anos])
axs[0,1].set_title('Brasil')

axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Perú', anos])
axs[1,1].set_title('Perú')

ymin = 0
ymax = 7000

#ravel: permite traer un array multidimensional a un array unidimensional
for ax in axs.ravel():
    ax.set_ylim(ymin, ymax)

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))
    ax.set_xlabel('Año')
    ax.set_ylabel('Nro de Inmigrantes')
    ax.grid()

plt.show()
```

### Desafio: visualizando datos de ventas de tiendas diferentes

```python
tiendas = ['A', 'B', 'C', 'D']
ventas_2022 = {'Ene': [100, 80, 150, 50],
    'Feb': [120, 90, 170, 60],
    'Mar': [150, 100, 200, 80],
    'Abr': [180, 110, 230, 90],
    'May': [220, 190, 350, 200],
    'Jun': [230, 150, 280, 120],
    'Jul': [250, 170, 300, 140],
    'Ago': [260, 180, 310, 150],
    'Sep': [240, 160, 290, 130],
    'Oct': [220, 140, 270, 110],
    'Nov': [400, 220, 350, 190],
    'Dec': [300, 350, 400, 250]}
```

```python
#creo el dataframe
df2 = pd.DataFrame(ventas_2022, index=tiendas)
df2
```

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 8))
fig.subplots_adjust(hspace=0.3, wspace=0.4)#ajusto la dimension horizontal high y width
fig.suptitle('ventas en el pediodo de enero a diciembre de 2022')

axs[0,0].plot(df2.loc['A'])
axs[0,0].set_title('Ventas en la tienda A')

axs[0,1].plot(df2.loc['B'])
axs[0,1].set_title('Ventas en la tienda B')

axs[1,0].plot(df2.loc['C'])
axs[1,0].set_title('Ventas en la tienda C')

axs[1,1].plot(df2.loc['D'])
axs[1,1].set_title('Ventas en la tienda D')

for ax in axs.flat:
    ax.set_xlabel('Mes')
    ax.set_ylabel('Nro de Ventas')

plt.show()
```

### 03- Personalizando graficos con Matplotlib

```python
#----------Personalizando visualizaciones con matplotlib----------
#lw: ancho de la linea
#marker: añade un punto en cada valor del grafico
#fontsize: tamaño de la letra
#loc: posicion del titulo, left-center-right

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'], lw=3) #marker='o')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))#remplaza el xticks de los años
ax.set_title('Inmigracion de Colombianos hacia canadá\n1980 - 2013', fontsize=18, loc='left')
ax.set_xlabel('Año', fontsize=14)
ax.set_ylabel('Nro de Inmigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.grid()
plt.show()
```

```python
#----------Alterando colores #1----------
#color: se cambia el color a la grafica
#linestyle: cambiar el estilo de linea del grid
#

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(datos_col['Año'], datos_col['Inmigrantes'], lw=3, color='g') #marker='o')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))#remplaza el xticks de los años
ax.set_title('Inmigracion de Colombianos hacia canadá\n1980 - 2013', fontsize=18, loc='left')
ax.set_xlabel('Año', fontsize=14)
ax.set_ylabel('Nro de Inmigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.grid(linestyle='--')
plt.show()
```

### 04- Conociendo la biblioteca Seaborn

```python
#----------Creando el primer grafico con seaborn----------

import seaborn as sns
```

```python
#para mantener un mismo tema
sns.set_theme()
```

```python
top_10 = df.sort_values('Total', ascending=False).head(10)
top_10
```

```python
ax = sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h')
ax.set(title='Top 10 de paises con mayor migracion hacia Canadá\n 1980 - 2013', xlabel='Nro de Inmigrantes', ylabel='Paises')

plt.show()
```

```python
#----------Personalizando las visualizaciones con seaborn----------

fig, ax = plt.subplots(figsize=(8,4))

ax = sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h')
ax.set_title('Top 10 de paises con mayor migracion hacia Canadá\n 1980 - 2013', loc='left', fontsize = 16)
ax.set_xlabel('Nro de Inmigrantes', fontsize=14)
ax.set_ylabel('')

plt.show()
```

```python
#----------Utilizando diversas palettes----------

def generar_grafico(palette):
    fig, ax = plt.subplots(figsize=(8,4))
    ax.set_frame_on(False)#quita el marco del grafico
    ax.xaxis.set_visible(False)#quita los ejes del grafico
    ax.tick_params(axis='y', which='both', size=0)
    ax = sns.barplot(data=top_10, x='Total', y=top_10.index, hue=top_10.index, orient='h', palette=palette, legend=False)
    ax.set_title('Top 10 de paises con mayor migracion hacia Canadá\n 1980 - 2013', loc='left', fontsize = 16)
    ax.set_xlabel('Nro de Inmigrantes', fontsize=14)
    ax.set_ylabel('')
    
    for i,j in enumerate(top_10['Total']):
        ax.text(j+20, i, str(j), color='black', fontsize = 8.5, ha='left', va='center')

    plt.show()
```

```python
#----------Explorando otros temas----------

def generar_grafico(palette):
    fig, ax = plt.subplots(figsize=(8,4))
    ax.set(xticklabels=[])
    ax = sns.barplot(data=top_10, x='Total', y=top_10.index, hue=top_10.index, orient='h', palette=palette, legend=False)
    ax.set_title('Top 10 de paises con mayor migracion hacia Canadá\n 1980 - 2013', loc='left', fontsize = 16)
    ax.set_xlabel('Nro de Inmigrantes', fontsize=14)
    ax.set_ylabel('')
    sns.despine()#sacar los bordes del grafico
    
    for i,j in enumerate(top_10['Total']):
        ax.text(j+20, i, str(j), color='black', fontsize = 8.5, ha='left', va='center')

    plt.show()
```

```python
sns.set_theme(style='darkgrid')
generar_grafico('tab10')
```

```python
sns.set_theme(style='whitegrid')
generar_grafico('tab10')
```

### 05- Graficos interactivos con Plotly

```python
#----------Creando nuestro primer grafico interactivo----------

import plotly.express as px
```

```python
fig = px.line(datos_col, x='Año', y='Inmigrantes')
fig.update_layout(width=600, height=400, xaxis={'tickangle': -45})

fig.show()
```

```python
#----------Añadiendo rotulos y modificando fuentes----------

fig = px.line(datos_col, x='Año', y='Inmigrantes', title='Inmigración de colobianos hacia canadá entre 1980 - 2013')
fig.update_layout(width=800, height=400, xaxis={'tickangle': -45}, xaxis_title='Año', yaxis_title='Nro de Inmigrantes', font_family='Arial', font_size=14, font_color='grey')

fig.show()
```

```python
#----------Alterando colores con plotly----------

fig = px.line(datos_col, x='Año', y='Inmigrantes', title='Inmigración de colobianos hacia canadá entre 1980 - 2013')
fig.update_traces(line_color='green', line_width=3)
fig.update_layout(width=800, height=400, xaxis={'tickangle': -45}, xaxis_title='Año', yaxis_title='Nro de Inmigrantes', font_family='Arial', font_size=14, font_color='grey')

fig.show()
```
