# Pandas: Transformacion y Manipulacion de datos

### 0: Problema de negocio

Se trata de un algoritmo capaz de analizar las caracteristicas de una propiedad -como comodidades, tamaño, ocupacion del espacio en un pediodo determinado y sugerir un precio a cobrar por tafifas diarias que garantice ganancias en momentos de alta demanda

### 01: Conociendo el problema

```python
import pandas as pd
```

```python
#leer archivo inportandolo
datos = pd.read_json('archivos_01/datos_hosting.json')
datos.head()
```

```python
#normalizar los datos del json
datos = pd.json_normalize(datos['info_inmuebles'])
datos.head()
```

### 02: Datos numericos

```python
#---------Retirando las listas---------

columnas = list(datos.columns)
columnas
```

```python
#a partir de la columna descripcion_local(contienen conjunto de datos) aplico metodo explode
datos = datos.explode(columnas[3:])
datos
```

```python
#resetear indices
#inplace:para actualizar en el mismo dataframe
#drop:para borrar el indice anterior
datos.reset_index(inplace=True, drop=True) 
datos.head()
```

```python
datos.info()
```

```python
#---------Convirtiendo datos numericos---------

import numpy as np
```

```python
#convirtiendo columna max_hospedes a entero
datos['max_hospedes'] = datos['max_hospedes'].astype(np.int64)
```

```python
datos.info()
```

```python
#creo una lista de columnas numericas de cantidad
col_numericas = ['cantidad_baños', 'cantidad_cuartos', 'cantidad_camas']

datos[col_numericas] = datos[col_numericas].astype(np.int64)
```

```python
datos.info()
```

```python
#convertir columna evaluacion_general a tipo decimal
datos['evaluacion_general'] = datos['evaluacion_general'].astype(np.float64)
```

```python
datos.info()
```

```python
#columna precio $110.00 para convertilo a float hay que remover el signo $
datos['precio'] = datos['precio'].astype(np.float64) #esto no funciona!!!!
```

```python
datos['precio']
```

```python
#---------Numeros en strings---------

#aplico una funcion lambda para retirar el signo $, si hay , sacarla y eliminar espacios
datos['precio'] = datos['precio'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
datos.info()
```

```python
#cambiar el tipo de datos de columna precio
datos['precio'] = datos['precio'].astype(np.float64)
```

```python
datos.info()
```

```python
#---------Transformacion en multiples columnas---------

datos[['cuota_deposito', 'cuota_limpieza']]
```

```python
#aplicar apply() en varias columnas(dataframe) tengo que usar applymap
datos[['cuota_deposito', 'cuota_limpieza']] = datos[['cuota_deposito', 'cuota_limpieza']].applymap(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
datos[['cuota_deposito', 'cuota_limpieza']] = datos[['cuota_deposito', 'cuota_limpieza']].astype(np.float64)
```

```python
datos.info()
```

### 03: Datos de texto

```python
#--------- El problema del texto ---------
#convierto objeto a string(str) y lo paso a minuscula
datos['descripcion_local'] = datos['descripcion_local'].str.lower()
```

```python
datos.head(3)
```

```python
#--------- Eliminando caracteres con Regex ---------
datos['descripcion_local'][3169]
```

```python
datos['descripcion_local'] = datos['descripcion_local'].str.replace('[^a-zA-Z0-9\-\']',' ', regex=True)
```

```python
#reemplazar los guinoes que no estan entre palabras por espacio
datos['descripcion_local'] = datos['descripcion_local'].str.replace('(?<!\w)-(?!\w)',' ', regex=True)
```

```python
datos['descripcion_local'].head()
```

```python
#--------- Tokenizacion de strings ---------
datos['descripcion_local'] = datos['descripcion_local'].str.split()
datos.head()
```

```python
#ademas de buscar la llave inicial y final, tambien las comillas dobles
datos['comodidades'] = datos['comodidades'].str.replace('\{|}|\"', ' ', regex=True)
```

```python
#le indico el separador porque es una , y no por defecto espacio
datos['comodidades'] = datos['comodidades'].str.split(',')
datos.head()
```

### 04: Datos de tiempo

```python
#--------- Transformando datos de tiempo ---------

dt_data = pd.read_json('archivos_01/inmuebles_disponibles.json')
dt_data.head()
```

```python
dt_data.info()
```

```python
#cambio la columna fecha de object a datetime
dt_data['fecha'] = pd.to_datetime(dt_data['fecha'])
```

```python
dt_data.info()
```

```python
dt_data.sample(10)
```

```python
#--------- Manipulando datos de tiempo ---------
#agrupar por meses
dt_data['fecha'].dt.strftime('%Y-%m')
```

```python
#agrupar por año y mes
subset = dt_data.groupby(dt_data['fecha'].dt.strftime('%Y-%m'))['lugar_disponible'].sum()
subset
```
