* Proyecto Desafío 1: Ventas Online  
Debido a una inestabilidad en el sitio web de la empresa, tuvimos problemas con los nombres de los clientes durante el guardado. Esto resultó en una columna de nombres de clientes con una combinación de letras, mayúsculas y minúsculas, números y otros caracteres.  

Sabiendo esto, manipula los textos de la columna *Cliente* para que el resultado sean los nombres de los clientes en letras minúsculas, con ausencia de caracteres especiales o números.

```python
datos.head()
```

```python
datos['Cliente']
```

```python
#minuscula
datos['Cliente'] = datos['Cliente'].str.lower()
```

```python
#eliminar caracteres especiales o numeros
#datos['Cliente'].str.replace('[^a-z ]', ' ', regex=True)#problema: elimina caracter acentuado
#mantiene minus, acentuados, espacios en blanco y elimina todas las ocurrencias no solo la primera
datos['Cliente'] = (
    datos['Cliente']
    .str.normalize('NFKD') #separar acentos
    .str.encode('ascii', errors='ignore') #elimina acentos
    .str.decode('utf-8')
    .str.replace(r'[^a-zA-Z ]', ' ', regex=True) #eliminar numeros y simbolos
    .str.lower()
    .str.replace(r'\\s+', ' ', regex=True) #convertir espacios multiples en uno solo
    .str.strip()
)
datos['Cliente']
```
